1. Use CountVectorizer to generate the Bag of Words representa on.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
```

```python
texts = [
    "Football is an exciting sport with talented players",
    "Technology makes learning easier and faster",
    "Delicious food makes people happy and healthy"
]
```

2. Use TfidfVectorizer to compute TF-IDF scores.

```python
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words:")
print("Features:", count_vectorizer.get_feature_names_out())
print(bow_matrix.toarray())
```

3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

features = tfidf_vectorizer.get_feature_names_out()
scores = tfidf_matrix.toarray()

print("\nTF-IDF Scores:")
for i, row in enumerate(scores):
    print(f"\nText {i + 1}: {texts[i]}")
    for word, score in zip(features, row):
        if score > 0:
            print(f"{word}: {score:.3f}")
```

```python
print("\nTop 3 Keywords from Each Text:")
for i, row in enumerate(scores):
    top_indices = row.argsort()[-3:][::-1]
    top_words = [(features[j], round(row[j], 3))
                 for j in top_indices]
    print(f"Text {i + 1}: {top_words}")
```

Q4. Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).

```python
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import nltk

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
```

```python
text1 = """Artificial Intelligence enables machines to learn from data.
AI helps in decision making and problem solving.
It is used in healthcare, robotics, and automation.
Machine learning improves predictions using experience."""

text2 = """Blockchain is a decentralized digital ledger.
It stores data securely across multiple computers.
Blockchain supports cryptocurrencies and smart contracts.
It improves transparency and prevents data tampering."""
```

1. Preprocess and tokenize both texts.

```python
stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = re.sub(r'[^a-zA-Z\s]', '', text.lower())
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    return tokens

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("Tokens of Text 1:", tokens1)
print("Tokens of Text 2:", tokens2)
```

2. Calculate: a. Jaccard Similarity using sets

```python
set1 = set(tokens1)
set2 = set(tokens2)

jaccard = len(set1 & set2) / len(set1 | set2)

print("\nJaccard Similarity:", round(jaccard, 4))
```

b. Cosine Similarity using TfidfVectorizer + cosine_similarity()

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(
    tfidf_matrix[0:1], tfidf_matrix[1:2]
)[0][0]

print("Cosine Similarity:", round(cosine, 4))
```

c. Analyze which similarity metric gives be er insights in your case.

```python
print("\nAnalysis:")
print("Jaccard measures overlap between unique words.")
print("Cosine measures similarity between TF-IDF vectors.")
```

Q5. Write a short review for a product or service.

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

1. Use TextBlob or VADER to find polarity & subjec vity for each review.

```python
reviews = [
    "This phone is amazing and the camera quality is excellent.",
    "The product is terrible and the battery life is very poor.",
    "The phone is okay and has average features.",
    "I love this phone because it is fast and easy to use.",
    "The service was disappointing and the staff was unhelpful."
]

positive_reviews = []
```

2. Classify reviews into Posi ve / Nega ve / Neutral.

```python
for review in reviews:
    blob = TextBlob(review)
    polarity = blob.sentiment.polarity
    subjectivity = blob.sentiment.subjectivity

    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("\nReview:", review)
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)
```

3. Create a word cloud using the wordcloud library for all posi ve reviews.

```python
if positive_reviews:
    positive_text = " ".join(positive_reviews)

    wordcloud = WordCloud(
        width=800,
        height=400,
        background_color="white"
    ).generate(positive_text)

    plt.figure(figsize=(10, 5))
    plt.imshow(wordcloud, interpolation="bilinear")
    plt.axis("off")
    plt.title("Word Cloud of Positive Reviews")
    plt.show()
else:
    print("No positive reviews available.")
```

Q6. Choose your own paragraph (~100 words) as training data.

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
text = """
Football is a popular sport played around the world.
Players work together as a team to score goals.
Training improves fitness, speed, and coordination.
A good player practices every day and learns new skills.
Teamwork and discipline are important in football.
Players pass the ball and defend their goal.
Coaches teach strategies to help the team win matches.
Football brings people together and creates excitement.
Every match teaches players patience and confidence.
Hard work and practice help players become better.
Fans support their favorite teams with great passion.
Football is a wonderful game of skill and teamwork.
"""
```

1. Tokenize text using Tokenizer() from keras.preprocessing.text

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1

# Create input sequences
input_sequences = []
for line in text.strip().split("."):
    sequence = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(sequence)):
        input_sequences.append(sequence[:i + 1])

# Pad sequences
max_len = max(len(seq) for seq in input_sequences)
input_sequences = pad_sequences(
    input_sequences, maxlen=max_len, padding="pre"
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]
```

2. Create input sequences and build a simple LSTM or Dense model

```python
model = Sequential([
    Embedding(total_words, 32, input_length=max_len - 1),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

# Train the model
model.fit(X, y, epochs=200, verbose=0)

print("Model training completed!")
```

3. Train the model and generate 2–3 new lines of text star ng from any seed word you provide.

```python
def generate_text(seed_text, next_words=10):
    for _ in range(next_words):
        sequence = tokenizer.texts_to_sequences([seed_text])[0]
        sequence = pad_sequences(
            [sequence], maxlen=max_len - 1, padding="pre"
        )
        predicted = model.predict(sequence, verbose=0)
        predicted_word_index = np.argmax(predicted, axis=1)[0]

        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break

        if not output_word:
            break

        seed_text += " " + output_word

    return seed_text
```

```python
# Generate 2-3 lines
print("\nGenerated Text:")
print(generate_text("football", 10))
print(generate_text("players", 10))
print(generate_text("teamwork", 10))
```
