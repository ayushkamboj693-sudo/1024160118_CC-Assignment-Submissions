```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
```

```python
corrected_temperature = temperature + 2

fahrenheit = (9/5) * temperature + 32

readings_above_32 = temperature[temperature > 32]

count_above_32 = np.sum(temperature > 32)

print("Original temperatures:", temperature)
print("Corrected temperatures (+2°C):", corrected_temperature)
print("Fahrenheit:", fahrenheit)
print("Readings above 32°C:", readings_above_32)
print("Number of readings above 32°C:", count_above_32)

print("\nExplanation:")
print("Vectorization applies operations to the entire array at once,")
print("while Boolean indexing efficiently selects values meeting a condition.")
print("Both avoid explicit for-loops and are generally faster and more concise.")
```

```python
import numpy as np

# Steps recorded by 4 users over 3 days
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps recorded
total_steps = np.sum(steps)

# b. Mean number of steps
mean_steps = np.mean(steps)

# c. Maximum and minimum values
maximum = np.max(steps)
minimum = np.min(steps)

# d. Total steps for each day (axis=0)
daily_totals = np.sum(steps, axis=0)

# e. Total steps for each user (axis=1)
user_totals = np.sum(steps, axis=1)

# f. User/day position containing maximum number of steps
max_position = np.unravel_index(np.argmax(steps), steps.shape)

# Display results
print("a. Total steps:", total_steps)
print("b. Mean steps:", mean_steps)
print("c. Maximum steps:", maximum)
print("   Minimum steps:", minimum)
print("d. Total steps for each day:", daily_totals)
print("e. Total steps for each user:", user_totals)
print("f. Maximum value position (user, day):", max_position)
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()
copied_array[0] = 500

print("Original:", original)
print("Copied array:", copied_array)

arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)

print("3 x 4 Matrix:\n", matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("Flatten:", flattened)
print("Ravel:", raveled)

raveled[0] = 100
print("After modifying ravel:")
print("Original matrix:\n", matrix)
print("Ravel:", raveled)

flattened[1] = 200
print("After modifying flatten:")
print("Original matrix:\n", matrix)
print("Flatten:", flattened)

print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("a. Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("\nb. X.T:")
print(XT)

XT_X = XT @ X
print("\nc. X.T @ X:")
print(XT_X)

inverse = np.linalg.inv(XT_X)
print("\nd. Inverse of X.T @ X:")
print(inverse)

beta = inverse @ XT @ y
print("\ne. Coefficients (beta):")
print(beta)

print("\nf. Coefficients represent the effect of sleep hours, activity level, and stress level on the assistance score.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\ng. Predicted assistance score:", predicted_score)
```
